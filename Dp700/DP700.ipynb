```python
# Databricks Utilities (dbutils) - Overview
# dbutils is a utility module available in Databricks notebooks that provides
# helper functions for working with object storage, chaining notebooks, 
# managing secrets, and configuring widgets.

# List all available dbutils modules
dbutils.help()
```

```python
# dbutils.fs - File System Utilities
# Provides utilities for working with FileSystems (DBFS, Unity Catalog Volumes, cloud storage)
print("=== File System Utilities (dbutils.fs) ===")
dbutils.fs.help()

print("\n" + "="*80 + "\n")

# dbutils.widgets - Widget Utilities
# Provides utilities for creating and managing notebook input widgets
print("=== Widget Utilities (dbutils.widgets) ===")
dbutils.widgets.help()
```

```python
dbutils.widgets.text('FILE',"")
FILE=dbutils.widgets.get('FILE')
```

```python
# Create a multi-select dropdown widget
dbutils.widgets.multiselect('MULTI_SELECT', 'option1', ['option1', 'option2', 'option3', 'option4'], 'Select one or more options')

# Get the selected values (returns comma-separated string)
selected_values = dbutils.widgets.get('MULTI_SELECT')
print(f"Selected values: {selected_values}")

# Convert to a list for easier use
selected_list = [v.strip() for v in selected_values.split(',')]
print(f"Selected as list: {selected_list}")

```

```python
dbutils.widgets.dropdown('Dropdown','OP1',['OP1','OP2','OP3'],'Select one or more options')
select_value=dbutils.widgets.get('Dropdown')
print(f"seleted_value:{select_value}")

list=[v.strip() for v in select_value.split(',')]
print(f"selected_list:{list}")
```
